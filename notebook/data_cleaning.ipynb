```python
df = df.sort_values("rank")
df = df.drop_duplicates(["theme", "text"], keep="first")
df.shape
```

```python
df["words"] = df["text"].str.split().str.len()
print(df["words"].describe())

df = df[df["words"] >= 55]
#break big text into smaller chunks
```
